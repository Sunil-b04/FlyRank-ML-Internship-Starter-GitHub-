## 2. The question: decision, action, cost of a wrong call

**Research question:** Which webpages should be prioritized for content review based on observable performance and content-related signals?

- **Decision:** Decide which pages a content team should review first when time and resources are limited.
- **Unit of analysis:** One pseudonymized content item (page); each row represents one page in the starter snapshot.
- **Output:** A ranked review list with a score, supporting signals, and possible review actions.
- **Who acts / action:** A content editor or SEO reviewer inspects the prioritized pages, verifies the evidence, and decides whether to update, expand, monitor, or leave a page unchanged.
- **Cost of a wrong recommendation:** Review time may be wasted on a page with little opportunity, while a page that deserves attention may be missed. The score must therefore be reviewed by a human and should not trigger automatic changes.

## 3. Quick look at the data (2–3 real numbers)

The code below loads the repository's anonymized starter CSV and calculates aggregate statistics. It does not display page IDs or client IDs. These numbers help establish the scale of the dataset and how common the starter decline proxy is; they do not prove that content refreshes cause recovery.

```python
from pathlib import Path
import pandas as pd
import subprocess, sys

# Find the CSV from the current working directory or its parents.
data_path = None
for base in [Path.cwd(), *Path.cwd().parents]:
    candidate = base / 'data' / 'raw' / 'content_refresh_anonymized.csv'
    if candidate.exists():
        data_path = candidate
        break

# In Colab, clone the repository if the dataset is not already present.
if data_path is None and 'google.colab' in sys.modules:
    repo_url = 'https://github.com/Sunil-b04/FlyRank-ML-Internship-Starter-GitHub-.git'
    repo_dir = Path('/content/FlyRank-ML-Internship-Starter-GitHub-')
    if not repo_dir.exists():
        subprocess.run(['git', 'clone', '--depth', '1', repo_url, str(repo_dir)], check=True)
    data_path = repo_dir / 'data' / 'raw' / 'content_refresh_anonymized.csv'

if data_path is None or not data_path.exists():
    raise FileNotFoundError('Starter CSV not found. Run this notebook from the repository or use Colab.')

df = pd.read_csv(data_path)
declining = df['trend_direction'].astype(str).str.lower().eq('down')
print(f'Pages in starter snapshot: {len(df):,}')
print(f'Distinct pseudonymized clients: {df["client_id"].nunique():,}')
print(f"Pages marked 'down' in the snapshot: {declining.sum():,} ({declining.mean():.1%})")

# Aggregate summary numbers only; do not display IDs or row-level data.
```

**Initial interpretation:** The starter snapshot is large enough to explore page-level prioritization across multiple clients. The `down` proportion is a current-window trend proxy defined in the supplied data, not a future outcome and not evidence that any specific update caused performance changes.
